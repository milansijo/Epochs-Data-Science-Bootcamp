```python
import sqlite3
conn=sqlite3.connect("/content/northwind.db")
```

```python
import pandas as pd
tables=pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)
tables
```

```python
#TOP 10 Selling Products
query="""
SELECT p.ProductName,
SUM(od.Quantity) AS
TotalQuantitySold FROM "Order Details" od
JOIN products p
ON od.ProductID=p.ProductID
GROUP BY p.ProductName
ORDER BY TotalQuantitySold DESC LIMIT 10;
"""
top_products = pd.read_sql(query,conn)
top_products
```

```python
#Top !0 Customer by Revenue
query="""
SELECT c.CompanyName,ROUND(SUM(od.UnitPrice*od.Quantity*(1-od.Discount)),2) AS Revenue
FROM Customers c
JOIN Orders o
ON c.CustomerID=o.CustomerID
JOIN "Order Details" od
ON o.OrderID=od.OrderID
GROUP BY c.CompanyName
ORDER BY Revenue DESC LIMIT 10;
"""
top_customer=pd.read_sql(query,conn)
top_customer
```

```python
#Monthly Sales Trend
query="""
SELECT strftime('%Y-%m',o.OrderDate) AS Month,
ROUND(SUM(od.UnitPrice*od.Quantity*(1-od.Discount)),2) AS Sales
FROM Orders o
JOIN "Order Details" od
ON o.OrderID=od.OrderID
GROUP BY Month
ORDER BY Month;
"""
monthly_sales=pd.read_sql(query,conn)
monthly_sales
```

```python
#Best Performing Category
query="""
SELECT c.CategoryName,
ROUND(SUM(od.UnitPrice*od.Quantity*(1-od.Discount)),2) AS Revenue
FROM Categories c
JOIN Products p
ON c.CategoryID=p.CategoryID
JOIN "Order Details" od
ON p.ProductID=od.ProductID
GROUP BY c.CategoryName
ORDER BY Revenue DESC;
"""
categories=pd.read_sql(query,conn)
categories
```

```python
#Customer Purchase Frequency
query="""
SELECT c.CompanyName,
COUNT(o.OrderID) AS NumberOfOrders
FROM Customers c
JOIN Orders o
ON c.customerID=o.CustomerID
GROUP BY c.CompanyName
ORDER BY NumberOfOrders DESC;
"""
purchase_frequency=pd.read_sql(query,conn)
purchase_frequency
```
